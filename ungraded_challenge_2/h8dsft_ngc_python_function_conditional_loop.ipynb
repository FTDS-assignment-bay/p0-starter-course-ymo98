```python
def number_of_word(sentence):
    '''Fungsi ini mengembalikan jumlah kata dalam suatu kalimat'''
    list_kata = sentence.split()
    return len(list_kata)

print(number_of_word('Hello World!'))
print(number_of_word('Once upon a time in a farway land'))
```

## SOAL 2

```python
def find_multiples(start_range, end_range, multiple):
    '''Function ini akan mengembalikan list angka dalam suatu range yang merupakan multiple suata angka'''
    list_multiple =[]

    while start_range <= end_range:
        if start_range % multiple == 0:
            list_multiple.append(start_range)
        start_range += 1
    
    return list_multiple

print(find_multiples(1,20,5))
print(find_multiples(1,50,10))
```

## SOAL 3

```python
def converter(score):
    '''This function takes input: `convert_to` and n-paired of student(s) name with its respective score'''
    if score <= 100 and score >=85:
        return {'letter_grade' : 'A', 'gpa' : 4}
    elif score <= 84.99 and score >=70:
        return {'letter_grade' : 'B', 'gpa' : 3}
    elif score <= 69.99 and score >=60:
        return {'letter_grade' : 'C', 'gpa' : 2.5}
    elif score <= 59.99 and score >=50:
        return {'letter_grade' : 'D', 'gpa' : 2}
    elif score <= 49.99 and score >=40:
        return {'letter_grade' : 'E', 'gpa' : 1.5}
    else:
        return {'letter_grade' : 'F', 'gpa' : 1}

def Grade_Converter(convert_to,**kwargs):
    for key, value in kwargs.items():
        dict_nilai = converter(value)
        if convert_to == 'gpa':
            kwargs[key] = dict_nilai['gpa']
        else:
            kwargs[key] = dict_nilai['letter_grade']
    return kwargs

print(Grade_Converter(convert_to='gpa', Adam=62, Faiz=91))
print(Grade_Converter(convert_to='letter_grade', Albert=90, Dwi=82, Syahdan=58, Veronica=84))
```

## SOAL 4

```python
def my_fibonacci(n):
    '''Function ini untuk membuat bilangan Fibonacci'''
    fibo = {0:0, 1:1}
    a = 2
    while a <= n:
        fibo[a] = fibo[a-1] + fibo[a-2]
        a += 1
    
    return fibo
```

```python
my_fibonacci(10)
```

```python
def return_n_fibo(x):
    '''Function ini untuk mengenbalikan n-th sequence dari bilangan Fibonacci'''
    bil = my_fibonacci(x)
    return bil[x]
```

```python
print(return_n_fibo(0))
print(return_n_fibo(4))
print(return_n_fibo(10))
```
