**TASK**

Create a library catalog system using OOP. 

The system should allow users to add books to the catalog, search for books by title or author, and remove books from the catalog.

**REQUIREMENTS**

1.Each book should have a title, author, and unique identifier

2.The system should provide the following functionality:

<li>Adding a book to the catalog.
<li>Searching for books by title or author.
<li>Removing a book from the catalog.
<li>Displaying all books in the catalog.

<br>
3.Do the exception handling

```python
# Membuat class Book
class Book:
    def __init__(self, id, title, author):
        self.id = id
        self.title = title
        self.author = author
    
    def detail_buku(self):
        return [self.id, self.title, self.author]
    
#Making a book input
A = Book(1,'A Mocking Bird','Anonymous')
B = Book(2,'Lima Sekawan','Enid')
C = Book(3,'Habit','James')

print(A.detail_buku())
print(B.detail_buku())
print(C.detail_buku())
```

```python
# Membuat class Library
class Library():
    def __init__(self, lib):
        self.lib = lib

    def add_book(self, a_book):
        self.lib.append(a_book.detail_buku())

    def remove_book(self,b_book):
        self.lib = [i for i in self.lib if i != b_book.detail_buku()]

    def search_book(self, book_title = None, book_author = None):
        i = 0
        try:
            while i <= len(self.lib):
                if self.lib[i][1] == book_title or self.lib[i][2] == book_author:
                    break
                else:
                    i += 1
            print(self.lib[i])
        except:
            print("Buku tidak ditemukan")

            
    def display_lib(self):
        print (self.lib)
```

```python
#Assigning a library
library_akabane = []
library_akabane = Library(library_akabane)

#Adding a book to library
library_akabane.add_book(A)
library_akabane.add_book(B)
library_akabane.add_book(C)
library_akabane.display_lib()
```

```python
#Removing a book to library
library_akabane.remove_book(B)
library_akabane.display_lib()
```

```python
library_akabane.search_book('A Mocking Bird')
library_akabane.search_book('Habit')
```
