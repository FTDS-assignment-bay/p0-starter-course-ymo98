### **Task 1**

Scrape coins price from https://coinmarketcap.com/ with following requirements:

The task is to get information about Price, Market Cap, Volume, and Coin supply.

Based on the Inspect element, we know that those information locate on:
- Price: ```<div class="sc-b3fc6b7-0 dzgUIj"><span>$63,726.87</span></div>``` 
- Market Cap: ```<span data-nosnippet="true" class="sc-11478e5d-1 jfwGHx">$1,258,491,567,492</span></p>```
- Volume: ```<p font-size="1" class="sc-71024e3e-0 bbHOdE font_weight_500" color="text" data-sensors-click="true">$40,855,511,985</p>```
- Coin Supply: ```<p font-size="1" font-weight="medium" color="text" data-sensors-click="true" class="sc-71024e3e-0 hhmVNu">19,756,443 BTC</p>```

````python
from bs4 import BeautifulSoup
from selenium import webdriver
import requests
import pandas as pd

page = requests.get("https://coinmarketcap.com/?page")
page

# Create an instance of the Chrome browser
driver = webdriver.Chrome()

# Open the URL
url = "https://coinmarketcap.com/?page=1"
driver.get(url)

# Extract the HTML content
html = driver.page_source

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(html, "html.parser")
````

````python
# Price
price = soup.find_all('div',{"class":"sc-b3fc6b7-0 dzgUIj"})
for p in price:
    print(p.get_text())
````

````python
# Market Cap
market_cap = soup.find_all('span',{"class":"sc-11478e5d-1 jfwGHx"})
for cap in market_cap:
    print(cap.get_text())
````

````python
# Volume
volume = soup.find_all('p',{"class":"sc-71024e3e-0 bbHOdE font_weight_500"})
for vol in volume:
    print(vol.get_text())
````

````python
#Coin Supply
coin_supply = soup.find_all('p',{"class":"sc-71024e3e-0 hhmVNu"})
for coins in coin_supply:
    print(coins.get_text())
````

````python
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time

# Create an instance of the Chrome browser
driver = webdriver.Chrome()

# Open the URL
url = "https://coinmarketcap.com/?page=1"
driver.get(url)

# Let the page load
time.sleep(10)

for _ in range(10):
    # Scroll down by a fixed amount (to simulate continuous scrolling)
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")

    # Wait for new coins to load
    time.sleep(10)

# The full page should be loaded now, extract the HTML content
html = driver.page_source

# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(html, "html.parser")

# Extract coin data
names = [name.get_text() for name in soup.find_all('p', {"class": "sc-71024e3e-0 ehyBa-d"})]
prices = [p.get_text() for p in soup.find_all('div', {"class": "sc-b3fc6b7-0 dzgUIj"})]
market_caps = [market_cap.get_text() for market_cap in soup.find_all('span', {"class": "sc-11478e5d-1 jfwGHx"})]
volumes = [volume.get_text() for volume in soup.find_all('p', {"class": "sc-71024e3e-0 bbHOdE font_weight_500"})]
coin_supplies = [coin_supply.get_text() for coin_supply in soup.find_all('p', {"class": "sc-71024e3e-0 hhmVNu"})]

# Check the length of each list
print(f"Names: {len(names)}, Prices: {len(prices)}, Market Caps: {len(market_caps)}, Volumes: {len(volumes)}, Coin Supplies: {len(coin_supplies)}")

# Find the maximum length among all lists
max_len = max(len(names), len(prices), len(market_caps), len(volumes), len(coin_supplies))

# To make sure all lists have the same length, filling missing values with None
names += [None] * (max_len - len(names))
prices += [None] * (max_len - len(prices))
market_caps += [None] * (max_len - len(market_caps))
volumes += [None] * (max_len - len(volumes))
coin_supplies += [None] * (max_len - len(coin_supplies))

# Create a DataFrame
data = pd.DataFrame({
    'Name': names,
    'Price': prices,
    'Market Cap': market_caps,
    'Volume': volumes,
    'Coin Supply': coin_supplies
})

# Display the scraped data
data
````

````python
# Initialize lists
prices_list = []
market_caps_list = []
volumes_list = []
coin_supplies_list = []

for i in range(50):
    url = f"https://coinmarketcap.com/?page={str(i+1)}"

    driver.get(url)

    # Page load
    time.sleep(5)

    for _ in range(10):
        # Scroll down
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")

        # Wait for new coins to load
        time.sleep(5)

    # Saving the page_source
    html = driver.page_source

    soup = BeautifulSoup(html, "html.parser")

    # Extract data for each coin on the page and extend to the lists
    prices_list += [price.get_text() if price else None for price in soup.find_all('div', {"class": "sc-b3fc6b7-0 dzgUIj"})]
    market_caps_list += [market_cap.get_text() if market_cap else None for market_cap in soup.find_all('span', {"class": "sc-11478e5d-1 jfwGHx"})]
    volumes_list += [volume.get_text() if volume else None for volume in soup.find_all('p', {"class": "sc-71024e3e-0 bbHOdE font_weight_500"})]
    coin_supplies_list += [coin_supply.get_text() if coin_supply else None for coin_supply in soup.find_all('p', {"class": "sc-71024e3e-0 hhmVNu"})]
````

````python
print(f"Prices: {len(prices_list)}, Market Caps: {len(market_caps_list)}, Volumes: {len(volumes_list)}, Coin Supplies: {len(coin_supplies_list)}")

# Find the maximum length among all lists
max_len = max(len(prices_list), len(market_caps_list), len(volumes_list), len(coin_supplies_list))

# Create a DataFrame from the lists
data = pd.DataFrame({
    'Price': prices_list,
    'Market Cap': market_caps_list,
    'Volume': volumes_list,
    'Coin Supply': coin_supplies_list
})

# Display the scraped data
data
````
