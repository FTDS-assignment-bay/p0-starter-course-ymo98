```sql
-----------------------------
--- Non Graded Challenge 5 --
-----------------------------

---------
-- DDL --
---------

-- Create table 'customers'
CREATE TABLE customers(
	customer_id INT PRIMARY KEY,
	customer_name VARCHAR(100),
	city VARCHAR (100)
);

-- Create table 'orders'
CREATE TABLE orders(
	order_id INT PRIMARY KEY,
	customer_id INT REFERENCES customers(customer_id),
	order_date DATE ,
	total_amount FLOAT
);

-- Insert data to 'customers' table
INSERT INTO customers(customer_id,customer_name,city)
VALUES
	(1, 'John Doe', 'New York'),
	(2, 'Jane Smith', 'Los Angeles'),
	(3, 'David Johnson', 'Chicago');

-- Insert data to 'orders' table
INSERT INTO orders(order_id,customer_id,order_date,total_amount)
VALUES
	(1, 1, '2022-01-10', 100.00),
	(2, 1, '2022-02-15', 150.00),
	(3, 2, '2022-03-20', 200.00),
	(4, 3, '2022-04-25', 50.00);

---------
-- DQL --
---------

-- Show table 
SELECT * FROM customers;
SELECT * FROM orders;

-- Query total number of orders made by each customer along 
-- with the customer name from 'customers' and 'orders' tables

SELECT
	customer_name,
	COUNT(*) AS total_orders
FROM customers
INNER JOIN orders
USING (customer_id)
GROUP BY customer_name;
```
