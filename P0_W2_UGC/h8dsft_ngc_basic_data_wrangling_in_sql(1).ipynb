### **How many columns in the `bikeshare_stations` table have `STRING` data?**

```python
query_string_col = """
SELECT
    COUNT(*) AS string_column
    -- EXCEPT(is_generated, generation_expression, is_stored, is_updatable)
  FROM
    `bigquery-public-data`.austin_bikeshare.INFORMATION_SCHEMA.COLUMNS
  WHERE
    table_name = 'bikeshare_stations' AND data_type = 'STRING'
"""

query_job4 = client.query(query_string_col)
result = query_job4.result().to_dataframe()
print(result)
```

### **Retrieve all trips where the `bike_id` is `not null` and the `duration` is greater than 1000 minutes**

```python
# Melihat tabel bikeshare_trips
query_job
```

```python
# Membuat query sesuai instruksi
query_5 = """
SELECT *
FROM `bigquery-public-data.austin_bikeshare.bikeshare_trips`
WHERE
  bike_id IS NOT NULL
  AND duration_minutes > 1000
"""

query_job5 = client.query(query_5)
result = query_job5.result().to_dataframe()
print(result)
```

### **Filter trips where the `trip_id` is "`Walk Up`" and the `end_station_name`  is "`Stolen`"**

```python
# Membuat query sesuai instruksi
query_6 = """
SELECT *
FROM `bigquery-public-data.austin_bikeshare.bikeshare_trips`
WHERE
  subscriber_type = 'Walk Up' AND end_station_name = 'Stolen'
"""

query_job6 = client.query(query_6)
result = query_job6.result().to_dataframe()
print(result)
```
