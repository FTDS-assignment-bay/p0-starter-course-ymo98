```python
#Exception Hanldling Contoh

def divide_numbers(a,b):
    try:
        if b==0:
            raise ValueError("Cannot divide by zero")
        return a/b
    except ValueError as e:
        print("Error: ", str(e))

divide_numbers(10,0)
```

```python
# Membuat class Book + Try-Except
class Book:
    def __init__(self, id, title, author):
        try:
            # Validate id
            if type(id) != int and id <= 0:
                raise ValueError("ID must be a positive integer")

            # Validate title and author
            if title == False or author == False:
                raise ValueError("Title and author cannot be empty")

            # Assign attributes if all validations pass
            self.id = id
            self.title = title
            self.author = author

        except ValueError as v:
            print(f"Error: {v}")
            # Set default values
            self.id = None
            self.title = "Unknown Title"
            self.author = "Unknown Author"
    
    def detail_buku(self):
        return [self.id, self.title, self.author]
    
#Making a book input
A = Book(1,'A Mocking Bird','Anonymous')
B = Book(2,'Lima Sekawan','Enid')
C = Book(3,'Habit','James')

print(A.detail_buku())
print(B.detail_buku())
print(C.detail_buku())
```

```python
# Membuat class Library
class Library():
    def __init__(self, lib):
        self.lib = lib

    def add_book(self,id,title,author):    
        a_book = Book(id,title,author)
        self.lib.append(a_book.detail_buku())
        
    def remove_book(self,b_book):
        self.lib = [i for i in self.lib if i != b_book.detail_buku()]

    def search_book(self, book_title = None, book_author = None):
        i = 0
        valid = False
        while i <= len(self.lib)-1:
            if self.lib[i][1] == book_title or self.lib[i][2] == book_author:
                print(self.lib[i])
                valid = True
                break
            i += 1
              
        if valid == False:
            print("The book is not found")
                    
    def display_lib(self):
        print (self.lib)
```

```python
#Assigning a library
library_akabane = []
library_akabane = Library(library_akabane)

#Adding a book to library
library_akabane.add_book(1,'A Mocking Bird','Anonymous')
library_akabane.add_book(2,'Lima Sekawan','Enid')
library_akabane.add_book(3,'Habit','James')
library_akabane.display_lib()
```

```python
#Removing a book to library
library_akabane.remove_book(B)
library_akabane.display_lib()
```

```python
#Search existing books
library_akabane.search_book('A Mocking Bird')
library_akabane.search_book('Habit')
#Search non-existent books
library_akabane.search_book('Yocia')
```
