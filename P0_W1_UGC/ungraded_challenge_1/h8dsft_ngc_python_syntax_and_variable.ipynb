**Transformasi ke dictionary**

Karakterisik dictionary > Bila ada dua atau lebih key yang sama, nilai dari key tersebut akan di-update dengan nilai dari key sama yang terakhir muncul di dalam dictionary.

```python
unique_customers_di = {}

#Count unique customer IDs
for a in customer_id:
    unique_customers_di[a] = a

#Number of unique customer IDs
unique_customer_count_di = len(unique_customers_di)

print(customer_id)
print(unique_customers_di)
print(unique_customer_count_di)
```
